```python
## Function that calculates the percentage of orders delivered later than promised
# for a specific delivery zone and time block
def late_order_percentage(promised_delivery_time, zone, time_block):

   # Get the list of actual delivery times for this zone and time block for a given promised delivery time
    times = delivery_times(zone, time_block, promised_delivery_time)

    total_orders = len(times)    # Count the total number of orders

# Count how many orders took longer than the promised time
    # (each True counts as 1, each False counts as 0)
    count = sum(x > promised_delivery_time for x in times)

 # If there are no orders, return 0.0 to avoid dividing by zero
    if total_orders == 0:
        return 0.0

    percentage = count / total_orders * 100  # Calculate the percentage of late orders

    return round(percentage, 2) # Return the percentage rounded to 2 decimal places
```

```python
def late_order_percentage3(promised_delivery_time, zone, time_block):

    def zone_time_combinations(zone, time_block):

      if zone == "all" and time_block == "all":
          return [(z, t) for z in ZONES for t in TIME_BLOCKS]

      elif zone == "all":
          return [(z, time_block) for z in ZONES]

      elif time_block == "all":
          return [(zone, t) for t in TIME_BLOCKS]

      else:
          return [(zone, time_block)]

    time_zone_combinations = zone_time_combinations(zone, time_block)

    total_orders = 0
    count = 0
    percentage = 0
    list_of_arrival_rate = []

    for combination in time_zone_combinations:

        t_zone, t_time_block = combination

        times = delivery_times(t_zone, t_time_block, promised_delivery_time)

        total_delivery_count = len(times)

        total_late_count = sum(x > promised_delivery_time for x in times)

        late_percentage = (total_late_count / total_delivery_count * 100) if total_delivery_count > 0 else 0

        total_orders = total_orders + len(times)

        count = count +sum(x > promised_delivery_time for x in times)

        #Adding an element to the list with 'percentage' as the first item, followed by zone and time block.
        list_of_arrival_rate.append((round(late_percentage, 2), t_zone, t_time_block))

    if total_orders > 0:
        percentage = count / total_orders * 100

    #Sorting the list contents on the basis of percentage of late arrivals being
    #the very first item in the tuple (index 0)
    list_of_arrival_rate.sort(reverse=True,key=lambda x: list(x)[0]) #lambda tells the sort function which element is the key to be sorted on
    return (round(percentage, 2),list_of_arrival_rate)
```

####Part II. Exploring the best promise (30 points)
#####PartA.(4 points) - Write a function to calculate the total cost per late order. Test your function using the figures in COSTS , and show the output.

#####Notes: Each late order is costly for Rosa.
#####- Direct cost due to refund issued to customers.
#####- Each late order leads to customer churn: the customer places fewer orders in the future, and each lost order is lost profit.
#####You may find the refund cost and number of churns per late order, and the profit margin per order in the dictionary COSTS .

```python
def impact_of_delay(promised_delivery_time, zone, time_block):

    def zone_time_combinations(zone, time_block):

      if zone == "all" and time_block == "all":
          return [(z, t) for z in ZONES for t in TIME_BLOCKS]

      elif zone == "all":
          return [(z, time_block) for z in ZONES]

      elif time_block == "all":
          return [(zone, t) for t in TIME_BLOCKS]

      else:
          return [(zone, time_block)]

    time_zone_combinations = zone_time_combinations(zone, time_block)
    rlist = []

    for combination in time_zone_combinations:
        t_zone, t_time_block = combination
        times = delivery_times(t_zone, t_time_block, promised_delivery_time)

        total_orders = len(times)
        total_delayed_orders = sum(x > promised_delivery_time for x in times)
        total_on_time_orders = total_orders - total_delayed_orders

        total_cost_of_delay = total_delayed_orders * COSTS["refund"] + total_delayed_orders * COSTS["churn_orders"] * COSTS["margin"]

        rlist.append((t_zone, t_time_block, total_orders, int(total_on_time_orders), int(total_delayed_orders),round(float(total_cost_of_delay),2)))

    return rlist
```

#####Part II - PartB.(26 points)

#####Rosa needs to decide the best promised delivery time for each pair of zone and time block. To do that, one could consider a range of promised delivery times in steps of 5 minutes (e.g., 5 minutes, 10 minutes, 15 minutes, ..., 50 minutes). For each promise in the list, one needs to calculate the net profit -- the profit collected from the customers less the late costs. Then she could choose the promise corresponding to the highest net profit.

#####Discuss how you can properly choose the range. Then, write a function to choose the best promise. Its inputs should be the zone, the time block, the list of promises to try, and the costs (e.g., COSTS ), so that you can reuse it in the web app in Part III.

####MY RESPONSE TO DISCUSS PART: To properly choose the best promised delivery time from a range of times in minutes, I would try to simulate the orders for each promised delivery time, calculate the net profit for that simulation and add it into a list. Finally, I'd sort the list in the descending order of the net profit which will put the best promise delivery time entry at the very top of the list (index 0).

```python
def impact_of_delay(promised_delivery_time, zone, time_block, costs):

    def zone_time_combinations(zone, time_block):

      if zone == "all" and time_block == "all":
          return [(z, t) for z in ZONES for t in TIME_BLOCKS]

      elif zone == "all":
          return [(z, time_block) for z in ZONES]

      elif time_block == "all":
          return [(zone, t) for t in TIME_BLOCKS]

      else:
          return [(zone, time_block)]

    time_zone_combinations = zone_time_combinations(zone, time_block)
    rlist = []

    for combination in time_zone_combinations:
        t_zone, t_time_block = combination
        times = delivery_times(t_zone, t_time_block, promised_delivery_time)

        total_orders = len(times)
        total_delayed_orders = sum(x > promised_delivery_time for x in times)
        total_on_time_orders = total_orders - total_delayed_orders
        profit_opportunity = total_orders * costs["margin"]
        total_cost_of_delay = total_delayed_orders * costs["refund"] + total_delayed_orders * costs["churn_orders"] * costs["margin"]
        net_profit = profit_opportunity - total_cost_of_delay

        rlist.append((promised_delivery_time,
                      t_zone,
                      t_time_block,
                      total_orders,
                      int(total_on_time_orders),
                      int(total_delayed_orders),
                      round(float(total_cost_of_delay),2),
                      round(float(net_profit),2)))

    return rlist

def best_promise(zone, time_block, promised_delivery_times, costs):
  blist = []
  for x in promised_delivery_times:
    #print("trying with promise time = ", x, " minutes")
    blist.append(*impact_of_delay(x, zone, time_block, costs))
  blist.sort(reverse=True, key=lambda x:list(x)[7])
  return (blist[0],blist) #returning best promise and all the other promises

```
